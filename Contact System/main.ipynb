# A SIMPLE CONTACT SYSTEM

```python
import json
```

```python
def create_contact():
    try:
        with open ("contact.json","r") as file:
            contact=json.load(file)
    except FileNotFoundError:
        contact={}

    name=input("Enter the name").title()
    number=input("Enter contact number").title()

    if name in contact:
        print("Name already Exists...")
    else:
        contact[name]=number

    with open ("contact.json","w") as file:
        json.dump(contact,file,indent=4)

    print("Contact Saved... :)")
```

```python
def search_contact():
    try:
        with open ("contact.json","r") as file:
            contact=json.load(file)
    except FileNotFoundError:
        print("Contact not Found!!!")
    
    name=input("Enter the name:").title()

    if name in contact:
        print(contact[name])
    else:
        print("name not found")
```

```python
search_contact()
```
