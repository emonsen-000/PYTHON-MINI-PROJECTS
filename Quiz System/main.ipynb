```python
def marking(q, marks,correct):
    if q==correct:
        marks+=5
    return marks
```

```python
def physics_quiz():
    marks=0

    print("=== PHYSICS QUIZ ===")

    q=input("Does light travel faster than sound?").lower()
    marks=marking(q,marks,"yes")

    q=input("Is force measured in Newtons?").lower()
    marks=marking(q,marks,"yes")

    q=input("Does a heavier object always fall faster than a lighter object in a vacuum?").lower()
    marks=marking(q,marks,"no")

    q=input("Is velocity a scalar quantity?").lower()
    marks=marking(q,marks,"no")

    q=input("Does electricity flow through a conductor?").lower()
    marks=marking(q,marks,"yes")

    print(f"Your total mark is {marks}")
```

```python
def chemistry_quiz():
    marks=0

    print("=== CHEMISTRY QUIZ ===")

    q=input("Is water made of hydrogen and oxygen?").lower()
    marks=marking(q,marks,"yes")

    q=input("Is oxygen a metal?").lower()
    marks=marking(q,marks,"no")

    q=input("Does a neutral atom have the same number of protons and electrons?").lower()
    marks=marking(q,marks,"yes")

    q=input("Does increasing temperature always decrease the rate of a chemical reaction?").lower()
    marks=marking(q,marks,"no")

    q=input("Is NaCl commonly known as table salt?").lower()
    marks=marking(q,marks,"yes")

    print(f"Your total mark is {marks}")
```
